```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
chronic_kidney_disease = fetch_ucirepo(id=336)

# data (as pandas dataframes)
X = chronic_kidney_disease.data.features
y = chronic_kidney_disease.data.targets

# metadata
print(chronic_kidney_disease.metadata)

# variable information
print(chronic_kidney_disease.variables)
```

```python
import pandas as pd

# Combinar features y target en un solo DataFrame
df = pd.concat([X, y], axis=1)

# Guardar como CSV
df.to_csv("chronic_kidney_disease.csv", index=False)

print("Dataset guardado como chronic_kidney_disease.csv")
```

```python
import pandas as pd

# Definir las columnas a procesar
cols = ["rbc", "pc"]

for col in cols:
    # Columna indicadora de valores faltantes
    Kidney1[f"{col}_missing"] = Kidney1[col].isnull().astype(int)

    # Normalización: normal = 0, abnormal = 1
    Kidney1[f"{col}_norm"] = Kidney1[col].map({"normal": 0, "abnormal": 1})

    # Si hay NaN, se mantienen como NaN en la columna normalizada

# Verificar resultado
print(Kidney1[["rbc", "rbc_missing", "rbc_norm"]].head(10))
print(Kidney1[["pc", "pc_missing", "pc_norm"]].head(10))
```

| sod       | Entero       | sodio                   |
| pot       | Entero       | potasio                 |
| hemo      | Entero       | hemoglobina             |
| pcv       | Entero       | volumen celular empaquetado |
| wbcc      | Entero       | recuento de glóbulos blancos |
| rbcc      | Entero       | recuento de glóbulos rojos |
| htn       | Categórico   | hipertensión            |
| dm        | Categórico   | diabetes mellitus       |
| cad       | Categórico   | enfermedad de las arterias coronarias |
