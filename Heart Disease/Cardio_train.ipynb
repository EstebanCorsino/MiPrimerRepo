```python
!pip install -q kagglehub

import kagglehub

import shutil

import os

# Descargar el dataset desde Kaggle

path = kagglehub.dataset_download(

    "sulianova/cardiovascular-disease-dataset"

)

print("Ruta de descarga original:")

print(path)

# Ruta destino dentro de /content

destino = "/content/cardiovascular"

# Copiar todo el dataset a /content/cardiovascular

shutil.copytree(

    path,

    destino,

    dirs_exist_ok=True

)

print("\nDataset copiado correctamente en:")

print(destino)

# Mostrar archivos descargados

print("\nArchivos disponibles:")

for root, dirs, files in os.walk(destino):

    for archivo in files:

        print(os.path.join(root, archivo))
```

```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
heart_disease = fetch_ucirepo(id=45)

# data (as pandas dataframes)
X = heart_disease.data.features
y = heart_disease.data.targets

# metadata
print(heart_disease.metadata)

# variable information
print(heart_disease.variables)
```

```python
import pandas as pd

# Save features (X) to a CSV file
X.to_csv('heart_disease_features.csv', index=False)
print('Features (X) saved to heart_disease_features.csv')
```

```python
# Save targets (y) to a CSV file
y.to_csv('heart_disease_targets.csv', index=False)
print('Targets (y) saved to heart_disease_targets.csv')
```

```python
import pandas as pd

# Concatenate X and y into a single DataFrame
combined_df = pd.concat([X, y], axis=1)

# Save the combined DataFrame to a CSV file
combined_df.to_csv('heart_disease_combined.csv', index=False)
print('Features and targets combined and saved to heart_disease_combined.csv')
```
