### Descripción de las Columnas del DataFrame

| Columna           | Tipo de Dato | Descripción                                                               | Valores posibles / Rango                                     |
| :---------------- | :----------- | :------------------------------------------------------------------------ | :----------------------------------------------------------- |
| `patientid`       | `int64`      | Identificador único del paciente.                                         | Entero                                                       |
| `age`             | `int64`      | Edad del paciente.                                                        | Entero (años)                                                |
| `gender`          | `int64`      | Género del paciente.                                                      | `0` = femenino, `1` = masculino                              |
| `chestpain`       | `int64`      | Tipo de dolor en el pecho.                                                | `0` = angina típica, `1` = angina atípica, `2` = dolor no anginal, `3` = asintomático |
| `restingBP`       | `int64`      | Presión arterial en reposo (mm Hg).                                       | Entero                                                       |
| `serumcholestrol` | `int64`      | Colesterol sérico (mg/dl).                                                | Entero                                                       |
| `fastingbloodsugar` | `int64`    | Azúcar en sangre en ayunas (> 120 mg/dl).                                 | `0` = falso, `1` = verdadero                                 |
| `restingrelectro` | `int64`      | Resultados electrocardiográficos en reposo.                               | `0` = normal, `1` = anomalía de la onda ST-T, `2` = hipertrofia ventricular izquierda probable |
| `maxheartrate`    | `int64`      | Frecuencia cardíaca máxima alcanzada.                                    | Entero                                                       |
| `exerciseangia`   | `int64`      | Angina inducida por el ejercicio.                                         | `0` = no, `1` = sí                                           |
| `oldpeak`         | `float64`    | Depresión del ST inducida por el ejercicio en relación con el reposo.     | Flotante                                                     |
| `slope`           | `int64`      | La pendiente del segmento ST máximo del ejercicio.                        | `0` = ascendente, `1` = plano, `2` = descendente             |
| `noofmajorvessels` | `int64`     | Número de vasos principales coloreados por fluoroscopia.                  | `0`, `1`, `2`, `3`, `4`                                      |
| `target`          | `int64`      | Presencia de enfermedad cardíaca. (Variable objetivo)                    | `0` = no hay enfermedad, `1` = hay enfermedad                |
