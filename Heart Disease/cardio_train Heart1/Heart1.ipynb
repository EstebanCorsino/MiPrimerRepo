Veamos si hay valores atipicos

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np

# Seleccionar solo columnas numéricas
num_cols = Heart1.select_dtypes(include=["int64","float64"]).columns

# 1. Resumen estadístico
print(Heart1[num_cols].describe())

# 2. Boxplots para cada variable numérica
for col in num_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=Heart1[col])
    plt.title(f"Boxplot de {col}")
    plt.show()

# 3. Detección con Z-score (umbral 3)
z_scores = np.abs(stats.zscore(Heart1[num_cols]))
outliers_z = (z_scores > 3).sum(axis=0)
print("Número de outliers por columna (Z-score):\n", outliers_z)

# 4. Detección con IQR
Q1 = Heart1[num_cols].quantile(0.25)
Q3 = Heart1[num_cols].quantile(0.75)
IQR = Q3 - Q1
outliers_iqr = ((Heart1[num_cols] < (Q1 - 1.5 * IQR)) |
                (Heart1[num_cols] > (Q3 + 1.5 * IQR))).sum()
print("Número de outliers por columna (IQR):\n", outliers_iqr)
```

## Descripción de las Columnas del Dataset Cardiaco

| Columna       | Tipo de Dato | Descripción                                                                      |
| :------------ | :----------- | :------------------------------------------------------------------------------- |
| **id**        | `int64`      | Identificador único del paciente.                                                |
| **age**       | `int64`      | Edad del paciente en días.                                                       |
| **gender**    | `int64`      | Género del paciente (1: mujer, 2: hombre).                                      |
| **height**    | `int64`      | Altura del paciente en centímetros.                                              |
| **weight**    | `float64`    | Peso del paciente en kilogramos.                                                 |
| **ap_hi**     | `int64`      | Presión arterial sistólica (valor más alto).                                     |
| **ap_lo**     | `int64`      | Presión arterial diastólica (valor más bajo).                                    |
| **cholesterol**| `int64`      | Nivel de colesterol (1: normal, 2: por encima de lo normal, 3: muy por encima). |
| **gluc**      | `int64`      | Nivel de glucosa (1: normal, 2: por encima de lo normal, 3: muy por encima).    |
| **smoke**     | `int64`      | Indica si el paciente fuma (0: no, 1: sí).                                       |
| **alco**      | `int64`      | Indica si el paciente consume alcohol (0: no, 1: sí).                            |
| **active**    | `int64`      | Indica si el paciente es físicamente activo (0: no, 1: sí).                     |
| **cardio**    | `int64`      | Variable objetivo: indica si el paciente tiene enfermedad cardiovascular (0: no, 1: sí). |
