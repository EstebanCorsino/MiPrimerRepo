En este caso para la variable ca, vamos a llenar los valores faltantes con la mediana

```python
# Calcular la mediana de la columna 'ca'
median_ca = Heart2["ca"].median()

# Rellenar los valores nulos con la mediana
Heart2["ca"] = Heart2["ca"].fillna(median_ca)

# Verificar que ya no existan valores nulos
print(Heart2["ca"].isnull().sum())
```

```python
# Imputar con el modo (valor más frecuente)
mode_thal = Heart2["thal"].mode()[0]
Heart2["thal"] = Heart2["thal"].fillna(mode_thal)

# Crear columna indicadora de nulos originales
Heart2["thal_missing"] = Heart2["thal"].isna().astype(int)

# Verificar
print(Heart2["thal"].isnull().sum())
print(Heart2["thal_missing"].value_counts())
```

Veamos si hay duplicados

```python
# Contar cuántas filas duplicadas hay
print("Número de duplicados:", Heart2.duplicated().sum())
```
