| Variable          | Tipo de Dato | Descripción                                                                  |
|:------------------|:-------------|:-----------------------------------------------------------------------------|
| `HeartDisease`    | `object`     | Indica si el paciente tiene alguna enfermedad cardíaca (Sí/No).               |
| `BMI`             | `float64`    | Índice de Masa Corporal del paciente.                                        |
| `Smoking`         | `object`     | Indica si el paciente ha fumado al menos 5 paquetes de cigarrillos en su vida (Sí/No). |
| `AlcoholDrinking` | `object`     | Indica si el paciente es un bebedor excesivo de alcohol (Sí/No).             |
| `Stroke`          | `object`     | Indica si el paciente ha tenido un derrame cerebral (Sí/No).                  |
| `PhysicalHealth`  | `float64`    | Número de días durante los últimos 30 en que la salud física no fue buena.    |
| `MentalHealth`    | `float64`    | Número de días durante los últimos 30 en que la salud mental no fue buena.   |
| `DiffWalking`     | `object`     | Indica si el paciente tiene dificultad para caminar o subir escaleras (Sí/No). |
| `Sex`             | `object`     | Sexo del paciente (Hombre/Mujer).                                            |
| `AgeCategory`     | `object`     | Categoría de edad del paciente (ej. 18-24, 25-29, etc.).                      |
| `Race`            | `object`     | Raza del paciente.                                                            |
| `Diabetic`        | `object`     | Indica si el paciente tiene diabetes (Sí/No/Borderline/Durante el embarazo). |
| `PhysicalActivity`| `object`     | Indica si el paciente realizó actividad física en los últimos 30 días (Sí/No). |
| `GenHealth`       | `object`     | Nivel de salud general percibido por el paciente (Excelente, Muy bueno, etc.). |
| `SleepTime`       | `float64`    | Horas de sueño promedio del paciente.                                        |
| `Asthma`          | `object`     | Indica si el paciente tiene asma (Sí/No).                                     |
| `KidneyDisease`   | `object`     | Indica si el paciente tiene enfermedad renal (Sí/No).                       |
| `SkinCancer`      | `object`     | Indica si el paciente ha tenido cáncer de piel (Sí/No).                      |
