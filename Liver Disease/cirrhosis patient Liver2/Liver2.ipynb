Dejamos la columna Drug con solo 3 variables, convirtiendo NaNN en nan

```python
# Reemplazar el texto "NaNN" por un verdadero NaN
Liver2["Drug"] = Liver2["Drug"].replace("NaNN", np.nan)

# Verificar los valores únicos después de la limpieza
print(Liver2["Drug"].unique())
```

Aplicamos one-hot encoding a Drug

```python
# Normalizar valores nulos
Liver2["Drug"] = Liver2["Drug"].replace("NaNN", np.nan)

# Crear columnas binarias
Liver2["Drug_Dpenicillamine"] = (Liver2["Drug"] == "D-penicillamine").astype(int)
Liver2["Drug_Placebo"] = (Liver2["Drug"] == "Placebo").astype(int)
Liver2["Drug_NA"] = Liver2["Drug"].isna().astype(int)

# Verificar tipos de datos
print(Liver2[["Drug_Dpenicillamine", "Drug_Placebo", "Drug_NA"]].dtypes)
```

Para la siguiente columna no conviene usar media ni mediana, porque son medidas numéricas y aquí hablamos de una variable categórica binaria, si usamos modas podriamos imputar y crear sesgos

```python
# Normalizar valores nulos
Liver2["Ascites"] = Liver2["Ascites"].replace("NaNN", np.nan)

# Codificación binaria
Liver2["Ascites"] = Liver2["Ascites"].map({"Y": 1, "N": 0})

# Columna indicadora de nulos
Liver2["Ascites_NA"] = Liver2["Ascites"].isna().astype(int)

# Verificar
print(Liver2[["Ascites", "Ascites_NA"]].head())
```

Aplicamos el mismo criterio a la columna Hepatomegaly que usamos anteriormente para Ascites

```python
# Normalizar valores nulos
Liver2["Hepatomegaly"] = Liver2["Hepatomegaly"].replace("NaNN", np.nan)

# Codificación binaria
Liver2["Hepatomegaly"] = Liver2["Hepatomegaly"].map({"Y": 1, "N": 0})

# Columna indicadora de nulos
Liver2["Hepatomegaly_NA"] = Liver2["Hepatomegaly"].isna().astype(int)

# Verificar
print(Liver2[["Hepatomegaly", "Hepatomegaly_NA"]].head())
```

Aplicamos el mismo criterio para Spider

```python
# Normalizar valores nulos
Liver2["Spiders"] = Liver2["Spiders"].replace("NaNN", np.nan)

# Codificación binaria
Liver2["Spiders"] = Liver2["Spiders"].map({"Y": 1, "N": 0})

# Columna indicadora de nulos
Liver2["Spiders_NA"] = Liver2["Spiders"].isna().astype(int)

# Verificar
print(Liver2[["Spiders", "Spiders_NA"]].head())
```

Vamos con Cholesterol

```python
Liver2["Cholesterol"].unique()
```

En este caso vemos mas recomendable usar la mediana porque no se ve afectada por valores extremos y refleja mejor el valor central de la distribución

```python
# Convertir a numérico
Liver2["Cholesterol"] = pd.to_numeric(Liver2["Cholesterol"], errors="coerce")

# Calcular la mediana
mediana_chol = Liver2["Cholesterol"].median()

# Imputar valores faltantes con la mediana
Liver2["Cholesterol"] = Liver2["Cholesterol"].fillna(mediana_chol)

# Verificar que ya no haya NaN
print(Liver2["Cholesterol"].isna().sum())
```

Copper aplicamos el mismo criterio que en Cholesterol

```python
Liver2["Copper"].unique()
```

```python
# Convertir a numérico
Liver2["Copper"] = pd.to_numeric(Liver2["Copper"], errors="coerce")

# Calcular la mediana
mediana_copper = Liver2["Copper"].median()

# Imputar valores faltantes con la mediana
Liver2["Copper"] = Liver2["Copper"].fillna(mediana_copper)

# (Opcional) Crear columna indicadora de nulos originales
Liver2["Copper_NA"] = Liver2["Copper"].isna().astype(int)

# Verificar
print(Liver2[["Copper", "Copper_NA"]].head())
```

Para imputar valores faltantes, la mediana es más recomendable que la media, porque esta variable suele tener valores extremos (outliers) en pacientes con enfermedad hepática.

```python
# Calcular la media
media_alk = Liver2["Alk_Phos"].mean()

# Imputar valores faltantes con la media
Liver2["Alk_Phos"] = Liver2["Alk_Phos"].fillna(media_alk)

# Verificar que ya no haya NaN
print(Liver2["Alk_Phos"].isna().sum())
```

Usamos el mismo criterio para Tryglicerides

```python
# Convertir Tryglicerides a numérico
Liver2["Tryglicerides"] = pd.to_numeric(Liver2["Tryglicerides"], errors="coerce")

# Crear columna indicadora de nulos originales
Liver2["Tryglicerides_NA"] = Liver2["Tryglicerides"].isna().astype(int)

# Verificar
print(Liver2[["Tryglicerides", "Tryglicerides_NA"]].head())
```

```python
Liver2["Platelets"].unique()
```

En este caso, la media es aceptable porque los datos están casi completos y los pocos nulos no afectarán demasiado.

```python
# Convertir a numérico
Liver2["Platelets"] = pd.to_numeric(Liver2["Platelets"], errors="coerce")

# Calcular la media
media_platelets = Liver2["Platelets"].mean()

# Imputar valores faltantes con la media
Liver2["Platelets"] = Liver2["Platelets"].fillna(media_platelets)

# Verificar
print(Liver2["Platelets"].head())
```

En este caso, la mediana sigue siendo más robusta, pero la diferencia con la media será mínima porque los nulos son casi inexistentes.

```python
# Convertir a numérico
Liver2["Prothrombin"] = pd.to_numeric(Liver2["Prothrombin"], errors="coerce")

# Calcular la mediana
mediana_prot = Liver2["Prothrombin"].median()

# Imputar valores faltantes con la mediana
Liver2["Prothrombin"] = Liver2["Prothrombin"].fillna(mediana_prot)

# Verificar
print(Liver2["Prothrombin"].head())
```

```python
Liver2["Stage"].unique()
```

Como los faltantes son muy pocos, imputar con la mediana es seguro y no distorsiona la distribución.

```python
# Convertir a numérico si aparece como object
Liver2["Stage"] = pd.to_numeric(Liver2["Stage"], errors="coerce")

# Calcular la mediana
mediana_stage = Liver2["Stage"].median()

# Imputar valores faltantes con la mediana
Liver2["Stage"] = Liver2["Stage"].fillna(mediana_stage)

# Verificar
print(Liver2["Stage"].head())
```
