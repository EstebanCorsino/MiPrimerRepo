| Columna | Tipo de Dato | Descripción |
|:---------------------------------------------------------------------------------------------------------------------|:-------------|:-------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| Patient No. | `int64` | Número de identificación único para cada paciente. |
| Age | `int64` | Edad del paciente. |
| Gender (Female=1, Male=2) | `int64` | Género del paciente, donde 1 es Femenino y 2 es Masculino. |
| Height | `int64` | Altura del paciente en centímetros. |
| Weight | `int64` | Peso del paciente en kilogramos. |
| Body Mass Index | `float64` | Índice de Masa Corporal (IMC) del paciente. |
| Waist Circumference | `float64` | Circunferencia de la cintura del paciente en centímetros. |
| Hip Circumference | `float64` | Circunferencia de la cadera del paciente en centímetros. |
| Systolic Blood Pressure | `int64` | Presión arterial sistólica del paciente. |
| Diastolic Blood Pressure | `int64` | Presión arterial diastólica del paciente. |
| Diyabetes Mellitus (No=0, Yes=1) | `int64` | Indicador de diabetes mellitus (0 = No, 1 = Sí). |
| Hypertension (No=0, Yes=1) | `int64` | Indicador de hipertensión (0 = No, 1 = Sí). |
| Hyperlipidemia (No=0, Yes=1) | `int64` | Indicador de hiperlipidemia (0 = No, 1 = Sí). |
| Metabolic syndrome (No=0, Yes=1) | `int64` | Indicador de síndrome metabólico (0 = No, 1 = Sí). |

| ALT | `float64` | Nivel de Alanina aminotransferasa. |
| ALP | `float64` | Nivel de Fosfatasa alcalina. |
| GGT | `float64` | Nivel de Gamma-glutamil transferasa. |
| LDH | `float64` | Nivel de Lactato deshidrogenasa. |
| Total Bilirubin | `float64` | Nivel de bilirrubina total. |
| Direct Bilirubin | `float64` | Nivel de bilirrubina directa. |
| Indirect Bilirubin | `float64` | Nivel de bilirrubina indirecta. |
| Albumin | `float64` | Nivel de albúmina. |
| Total Protein | `float64` | Nivel de proteínas totales. |
| Urea | `float64` | Nivel de urea. |
| Creatinine Clearance | `float64` | Nivel de aclaramiento de creatinina. |
| Gamma Globulin | `float64` | Nivel de Gamma Globulina. |
| Cholesterol | `float64` | Nivel de colesterol. |
| Triglycerides | `float64` | Nivel de triglicéridos. |
| LDL | `float64` | Nivel de colesterol LDL (lipoproteínas de baja densidad). |
| HDL | `float64` | Nivel de colesterol HDL (lipoproteínas de alta densidad). |
| VLDL | `float64` | Nivel de colesterol VLDL (lipoproteínas de muy baja densidad). |
| Uric Acid | `float64` | Nivel de ácido úrico. |
| White Blood Cells | `float64` | Recuento de glóbulos blancos. |

| Hemoglobin | `float64` | Nivel de hemoglobina. |

| INR | `float64` | Razón Normalizada Internacional. |
| Ferritin | `float64` | Nivel de ferritina. |
| Glucose | `float64` | Nivel de glucosa. |
| BUN | `float64` | Nivel de Nitrógeno Ureico en Sangre (BUN). |

| Hemoglobin - A1C | `float64` | Nivel de Hemoglobina A1C. |
| Steatosis | `int64` | Grado de esteatosis hepática. |
| Activity | `int64` | Grado de actividad inflamatoria hepática. |
| Fibrosis | `int64` | Grado de fibrosis hepática. |

| Type of Disease (Mild illness=1, Severe illness=2) | `int64` | Tipo de enfermedad (1 = Enfermedad leve, 2 = Enfermedad grave). |
