```python
import pandas as pd

Pulmo3 = pd.read_csv(os.path.join(path, "dataset.csv"))
print(Pulmo3.head())
```

```python
Pulmo3.info()
```

```python
# Contar cuántas filas duplicadas hay
print("Número de duplicados:", Pulmo3.duplicated().sum())
```

```python
Pulmo3.shape
```

```python
# Calcular la media de MWT1 (ignorando los nulos)
mean_mwt1 = Pulmo3["MWT1"].mean()

# Rellenar los valores faltantes con la media
Pulmo3["MWT1"] = Pulmo3["MWT1"].fillna(mean_mwt1)

# Verificar que ya no haya nulos
print("Nulos en MWT1 después de imputar:", Pulmo3["MWT1"].isnull().sum())
```

```python
# Calcular la media de MWT1 (ignorando los nulos)
mean_mwt1 = Pulmo3["MWT2"].mean()

# Rellenar los valores faltantes con la media
Pulmo3["MWT2"] = Pulmo3["MWT2"].fillna(mean_mwt1)

# Verificar que ya no haya nulos
print("Nulos en MWT2 después de imputar:", Pulmo3["MWT2"].isnull().sum())
```

```python
# Calcular la media de MWT1Best (ignorando los nulos)
mean_mwt1 = Pulmo3["MWT1Best"].mean()

# Rellenar los valores faltantes con la media
Pulmo3["MWT1Best"] = Pulmo3["MWT1Best"].fillna(mean_mwt1)

# Verificar que ya no haya nulos
print("Nulos en MWT1Best después de imputar:", Pulmo3["MWT1Best"].isnull().sum())
```

Normalizamos

```python
# Valores únicos en la columna 'COPDSEVERITY'
print(Pulmo3["COPDSEVERITY"].unique())
```

```python
# One-Hot Encoding para COPDSEVERITY
Pulmo3 = pd.get_dummies(Pulmo3, columns=["COPDSEVERITY"], prefix="COPDSEVERITY")

# Convertir las nuevas columnas a enteros (0 y 1)
Pulmo3[Pulmo3.filter(like="COPDSEVERITY_").columns] = Pulmo3.filter(like="COPDSEVERITY_").astype(int)

# Verificar resultado
for col in Pulmo3.filter(like="COPDSEVERITY_").columns:
    print(col, Pulmo3[col].unique(), Pulmo3[col].dtype)
```

```python
# Guardar Pulmo1 en un archivo CSV
Pulmo3.to_csv("Pulmo3.csv", index=False)

print("Archivo Pulmo3.csv descargado correctamente")
```

### Descripción de las Columnas del Dataset `Pulmo3`

A continuación, se presenta una descripción de cada columna en el DataFrame `Pulmo3`, basada en los nombres de las columnas y los tipos de datos observados.

| Columna                    | Tipo de Datos | Descripción                                                                                                                                                                             |
|:---------------------------|:--------------|:----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| `Unnamed: 0`               | `int64`       | Columna de índice o un identificador de fila original, a menudo generada al guardar/cargar CSVs. Puede ser redundante.                                                                    |
| `ID`                       | `int64`       | Identificador único para cada paciente o registro.                                                                                                                                      |
| `AGE`                      | `int64`       | Edad del paciente en años.                                                                                                                                                              |
| `PackHistory`              | `float64`     | Historial de tabaquismo del paciente, posiblemente en paquetes-año (packs-years).                                                                                                       |
| `MWT1`                     | `float64`     | Distancia caminada en la primera prueba de marcha de 6 minutos (6MWT).                                                                                                                    |
| `MWT2`                     | `float64`     | Distancia caminada en la segunda prueba de marcha de 6 minutos (6MWT).                                                                                                                   |
| `MWT1Best`                 | `float64`     | La mejor distancia caminada entre MWT1 y MWT2.                                                                                                                                          |
| `FEV1`                     | `float64`     | Volumen espiratorio forzado en 1 segundo (medida de función pulmonar).                                                                                                                  |
| `FEV1PRED`                 | `float64`     | Porcentaje predicho de FEV1 (FEV1 como porcentaje del valor esperado).                                                                                                                  |
| `FVC`                      | `float64`     | Capacidad vital forzada (medida de función pulmonar).                                                                                                                                   |
| `FVCPRED`                  | `int64`       | Porcentaje predicho de FVC (FVC como porcentaje del valor esperado).                                                                                                                    |
| `CAT`                      | `int64`       | Puntuación del Cuestionario de Evaluación de la EPOC (CAT score), mide el impacto de la EPOC en la vida del paciente.                                                                    |
| `HAD`                      | `float64`     | Puntuación de la Escala de Ansiedad y Depresión Hospitalaria (Hospital Anxiety and Depression scale).                                                                                    |
| `SGRQ`                     | `float64`     | Puntuación del Cuestionario Respiratorio de St. George (St. George's Respiratory Questionnaire), evalúa el estado de salud de los pacientes con enfermedades respiratorias.                 |
| `AGEquartiles`             | `int64`       | Cuartil de edad del paciente.                                                                                                                                                           |
| `copd`                     | `int64`       | Indicador numérico de la presencia o gravedad de la EPOC.                                                                                                                               |
| `gender`                   | `int64`       | Género del paciente (e.g., 0 para femenino, 1 para masculino).                                                                                                                          |
| `smoking`                  | `int64`       | Estado de fumador del paciente (e.g., 0 para no fumador, 1 para ex-fumador, 2 para fumador actual).                                                                                     |
| `Diabetes`                 | `int64`       | Indicador de si el paciente tiene diabetes (e.g., 0 para no, 1 para sí).                                                                                                                |
| `muscular`                 | `int64`       | Indicador de si el paciente tiene alguna condición muscular (e.g., 0 para no, 1 para sí).                                                                                               |
| `hypertension`             | `int64`       | Indicador de si el paciente tiene hipertensión (e.g., 0 para no, 1 para sí).                                                                                                            |
| `AtrialFib`                | `int64`       | Indicador de si el paciente tiene fibrilación auricular (e.g., 0 para no, 1 para sí).                                                                                                   |
| `IHD`                      | `int64`       | Indicador de si el paciente tiene enfermedad isquémica del corazón (e.g., 0 para no, 1 para sí).                                                                                        |
| `COPDSEVERITY_MILD`        | `int64`       | Columna dummy: 1 si la severidad de EPOC es MILD, 0 en otro caso.                                                                                                                       |
| `COPDSEVERITY_MODERATE`    | `int64`       | Columna dummy: 1 si la severidad de EPOC es MODERATE, 0 en otro caso.                                                                                                                   |
| `COPDSEVERITY_SEVERE`      | `int64`       | Columna dummy: 1 si la severidad de EPOC es SEVERE, 0 en otro caso.                                                                                                                     |
| `COPDSEVERITY_VERY SEVERE` | `int64`       | Columna dummy: 1 si la severidad de EPOC es VERY SEVERE, 0 en otro caso.                                                                                                                |
