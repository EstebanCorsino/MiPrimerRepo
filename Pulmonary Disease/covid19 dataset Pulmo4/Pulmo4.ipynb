```python
# cargar la base de datos
Pulmo4 = pd.read_csv("/content/Covid Data.csv")
```

```python
# Contar cuántas filas duplicadas hay
print("Número de duplicados:", Pulmo4.duplicated().sum())
```

```python
# Eliminar filas duplicadas y quedarte con la primera ocurrencia
Pulmo4 = Pulmo4.drop_duplicates()

# Verificar nuevamente
print("Número de duplicados después de limpiar:", Pulmo4.duplicated().sum())
```

```python
Pulmo4.shape
```

Normalizamos

```python
# Valores únicos en la columna 'COPDSEVERITY'
print(Pulmo4["DATE_DIED"].unique())
```

Normalizamos las fechas

```python
# 1. Reemplazar valores inválidos como '9999-99-99' por NaN
Pulmo4["DATE_DIED"] = Pulmo4["DATE_DIED"].replace("9999-99-99", np.nan)

# 2. Convertir a tipo datetime (día/mes/año)
Pulmo4["DATE_DIED"] = pd.to_datetime(Pulmo4["DATE_DIED"], format="%d/%m/%Y", errors="coerce")

# 3. Crear columna numérica: días desde la fecha mínima válida
Pulmo4["DATE_DIED_num"] = (Pulmo4["DATE_DIED"] - Pulmo4["DATE_DIED"].min()).dt.days

# 4. Normalización Min-Max al rango [0,1]
Pulmo4["DATE_DIED_norm"] = (Pulmo4["DATE_DIED_num"] - Pulmo4["DATE_DIED_num"].min()) / \
                           (Pulmo4["DATE_DIED_num"].max() - Pulmo4["DATE_DIED_num"].min())

# 5. Verificar resultado
print(Pulmo4[["DATE_DIED", "DATE_DIED_num", "DATE_DIED_norm"]].head())
```

```python
# Guardar Pulmo1 en un archivo CSV
Pulmo4.to_csv("Pulmo4.csv", index=False)

print("Archivo Pulmo4.csv descargado correctamente")
```
